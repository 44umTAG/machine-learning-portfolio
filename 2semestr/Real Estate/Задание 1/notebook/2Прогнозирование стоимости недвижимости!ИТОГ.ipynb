# Этап 2. Предварительная обработка данных 

## 2.1 Работа с пропусками

### Первичная очистка данных

Перед непосредственной обработкой пропущенных значений выполняется
первичная очистка исходного набора данных.

```python
df_clean = df.copy()

# Сохраняем оба типа объявлений: продажу (listing_type=1) и аренду (listing_type=2).
df_clean.drop(columns=['furnished', 'type'], inplace=True, errors='ignore')
df_clean = df_clean.drop_duplicates().copy()
df_clean = df_clean.dropna(subset=['price', 'listing_type']).copy()
df_clean = df_clean[df_clean['price'] > 0].copy()

# Приводим все цены к TRY (турецкая лира), чтобы модель не смешивала разные валюты.
# Коэффициенты ориентировочные и вынесены в одно место для удобного обновления под дату данных.
CURRENCY_TO_TRY = {'TRY': 1.0, 'TL': 1.0, '₺': 1.0,
                   'USD': 35.0, '$': 35.0, 'EUR': 38.0, '€': 38.0,
                   'GBP': 45.0, '£': 45.0}
currency = df_clean['price_currency'].astype(str).str.strip().str.upper()
rates = currency.map(CURRENCY_TO_TRY)
unknown_currency = sorted(currency[rates.isna()].unique().tolist())
if unknown_currency:
    print('ВНИМАНИЕ: неизвестные валюты исключены:', unknown_currency)
df_clean = df_clean.loc[rates.notna()].copy()
df_clean['price'] = df_clean['price'].astype(float) * rates.loc[df_clean.index]
df_clean['price_currency'] = 'TRY'

# Проверяем допустимые типы объявлений, не удаляя аренду.
df_clean = df_clean[df_clean['listing_type'].isin([1, 2])].copy()
print('Продажа:', (df_clean['listing_type'] == 1).sum())
print('Аренда:', (df_clean['listing_type'] == 2).sum())
print(f'Размер данных после очистки: {df_clean.shape}')
```

## 2.2 Обработка новой

```python
df_proc = df_clean.copy()
```

### Винзоризация

```python
# Границы выбросов считаем отдельно для продажи и аренды:
# месячная аренда и полная стоимость объекта имеют разные масштабы.
price_bounds = {}
for listing_type, group in df_proc.groupby('listing_type'):
    price_bounds[listing_type] = (group['price'].quantile(0.01), group['price'].quantile(0.99))
    print(f"Тип {listing_type}: цена {price_bounds[listing_type][0]:,.0f}–{price_bounds[listing_type][1]:,.0f} TRY")
```

```python
df_proc = pd.concat([
    group[(group['price'] >= price_bounds[listing_type][0]) &
          (group['price'] <= price_bounds[listing_type][1])]
    for listing_type, group in df_proc.groupby('listing_type')
], ignore_index=True)
```

На этапе обработки выбросов я применила винзоризацию по целевой переменной — отсекла 1% самых дешёвых и 1% самых дорогих объектов. Диапазон цен сузился с исходного до 65 000–3 600 000 лир, что позволило убрать экстремальные выбросы, не теряя при этом значимую часть данных — мы потеряли лишь около 4% выборки. 

### Логарифмирование сильно скошенных признаков

```python
df_proc['log_price'] = np.log1p(df_proc['price'])
df_proc['log_size'] = np.log1p(df_proc['size'])
```

На этом этапе я превратила текстовые и категориальные признаки в числовые, пригодные для обучения модели. Из столбца с этажом я извлекла четыре новых признака: три бинарных флага для нестандартных уровней — пентхаус, садовый этаж и цокольный/входной уровень — и один числовой признак с номером этажа. Это важно, потому что пентхаусы и цокольные квартиры имеют систематически отличающиеся цены, и модель должна это учитывать.

Далее я сгруппировала редкие категории в признаках типа недвижимости и типа отопления: оставила только топ-5 и топ-6 наиболее частых категорий соответственно, а всё остальное объединила в категорию "Other". Это снижает размерность и предотвращает переобучение на редких классах.

После этого применила One-Hot Encoding к четырём категориальным признакам — сгруппированным типу недвижимости, типу отопления, валюте и городу — с параметром drop_first=True для избежания мультиколлинеарности.

## 2.4 Работа с Feature Engineering

```python
X_feat = X_sale_enc.copy()

if 'district' in X_feat.columns:
    X_feat.drop(columns=['district'], inplace=True)

X_feat['size_per_room'] = X_feat['size'] / (X_feat['rooms'] + 1)
X_feat['is_new_building'] = (X_feat['building_age_num'] <= 5).astype(int)
X_feat['floor_ratio'] = X_feat['floor_num'] / (X_feat['total_floors_num'] + 1)
X_feat['has_living_room'] = (X_feat['living_rooms'] > 0).astype(int)

corr_matrix = X_feat.select_dtypes(include=['number']).corr()

high_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) > 0.8:
            high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j]))

cols_to_drop = set()
for col1, col2, val in high_corr:
    if val > 0.85:
        if X_feat[col1].nunique() < X_feat[col2].nunique():
            cols_to_drop.add(col1)
        else:
            cols_to_drop.add(col2)
            
X_feat = X_feat.dropna(subset=['rooms', 'size_per_room'])
print(f"Размер после удаления NaN: {X_feat.shape}")

X_feat = X_feat.drop(columns=['log_price', 'log_size'], errors='ignore')
```

На этом этапе я создала четыре новых информативных признака без утечки данных: площадь на комнату, флаг нового здания, отношение этажа к общему числу этажей и флаг наличия гостиной. Эти признаки помогают модели уловить нелинейные зависимости, которые не видны в исходных данных.

Затем я провела анализ мультиколлинеарности — нашла пары признаков с высокой корреляцией (более 0.8) и удалила менее информативные из пар с корреляцией выше 0.85. Это важно, потому что мультиколлинеарность дестабилизирует линейные модели и раздувает дисперсию оценок.

В завершение удалила строки с пропусками в ключевых признаках и вспомогательные логарифмированные столбцы. Итоговая матрица признаков содержит 173 994 объекта и 105 столбцов, полностью готовых к масштабированию и обучению модели.

```python
# Отдельная модель аренды: та же подготовка признаков и тот же набор столбцов,
# но обучение только на listing_type=2. Цена аренды не смешивается с ценой продажи.
df_rent = df_proc[df_proc['listing_type'] == 2].copy().reset_index(drop=True)
if len(df_rent) < 30:
    print("Недостаточно строк аренды для надёжного обучения:", len(df_rent))
else:
    X_rent = df_rent.drop(columns=['price', 'listing_type', 'log_price'], errors='ignore').copy()
    X_rent['city'] = X_rent['address'].astype(str).str.split('/').str[0].str.strip()
    X_rent['district'] = X_rent['address'].astype(str).str.split('/').str[1].str.strip().fillna('Unknown')
    X_rent['rooms'] = pd.to_numeric(X_rent['room_count'].astype(str).str.extract(r'^(\d+)')[0], errors='coerce')
    X_rent['living_rooms'] = pd.to_numeric(X_rent['room_count'].astype(str).str.extract(r'\+(\d+)')[0], errors='coerce').fillna(0)
    X_rent['building_age_num'] = pd.to_numeric(X_rent['building_age'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')
    X_rent['total_floors_num'] = pd.to_numeric(X_rent['total_floor_count'].astype(str).str.extract(r'^(\d+)')[0], errors='coerce')
    X_rent['is_penthouse'] = X_rent['floor_no'].astype(str).str.contains('Çatı|En Üst', na=False).astype(int)
    X_rent['is_garden'] = X_rent['floor_no'].astype(str).str.contains('Bahçe', na=False).astype(int)
    X_rent['is_ground'] = X_rent['floor_no'].astype(str).str.contains('Giriş|Zemin|Kot|Yüksek', na=False).astype(int)
    X_rent['floor_num'] = pd.to_numeric(X_rent['floor_no'].astype(str).str.extract(r'^(\d+)')[0], errors='coerce').fillna(0)
    top_subtypes_rent = ['Daire', 'Villa', 'Müstakil Ev', 'Rezidans', 'Yazlık']
    X_rent['sub_type_grouped'] = X_rent['sub_type'].where(X_rent['sub_type'].isin(top_subtypes_rent), 'Other')
    top_heating_rent = ['Kombi (Doğalgaz)', 'Klima', 'Merkezi Sistem (Isı Payı Ölçer)', 'Merkezi Sistem', 'Kalorifer (Doğalgaz)', 'Soba (Kömür)']
    X_rent['heating_grouped'] = X_rent['heating_type'].where(X_rent['heating_type'].isin(top_heating_rent), 'Other')
    X_rent = pd.get_dummies(X_rent, columns=['sub_type_grouped', 'heating_grouped', 'price_currency', 'city'],
                            drop_first=True, dtype=int)
    X_rent.drop(columns=['address', 'room_count', 'building_age', 'total_floor_count', 'floor_no',
                         'sub_type', 'heating_type', 'start_date', 'listing_date', 'tom', 'district'],
                inplace=True, errors='ignore')
    X_rent['size_per_room'] = X_rent['size'] / (X_rent['rooms'] + 1)
    X_rent['is_new_building'] = (X_rent['building_age_num'].fillna(10) <= 5).astype(int)
    X_rent['floor_ratio'] = X_rent['floor_num'] / (X_rent['total_floors_num'].fillna(5) + 1)
    X_rent['has_living_room'] = (X_rent['living_rooms'] > 0).astype(int)
    X_rent.drop(columns=['log_size'], errors='ignore', inplace=True)
    y_rent = df_rent['price'].reset_index(drop=True)
    X_rent = X_rent.reset_index(drop=True)
    X_rent = X_rent.select_dtypes(include=['number'])
    Xr_train, Xr_test, yr_train, yr_test = train_test_split(X_rent, y_rent, test_size=0.2, random_state=42)
    Xr_train, Xr_test = Xr_train.align(Xr_test, join='left', axis=1, fill_value=0)
    rent_imputer = SimpleImputer(strategy='median')
    Xr_train_imp = pd.DataFrame(rent_imputer.fit_transform(Xr_train), columns=Xr_train.columns)
    Xr_test_imp = pd.DataFrame(rent_imputer.transform(Xr_test), columns=Xr_test.columns)
    rent_model = RandomForestRegressor(n_estimators=200, max_depth=20, min_samples_leaf=3,
                                       random_state=42, n_jobs=-1)
    rent_model.fit(Xr_train_imp, yr_train)
    rent_pred = rent_model.predict(Xr_test_imp)
    print("Качество модели аренды (TRY за период объявления):")
    print('MAE:', round(mean_absolute_error(yr_test, rent_pred), 2))
    print('RMSE:', round(np.sqrt(mean_squared_error(yr_test, rent_pred)), 2))
    print('R²:', round(r2_score(yr_test, rent_pred), 4))
    joblib.dump(rent_model, 'rent_model.pkl')
    joblib.dump(rent_imputer, 'rent_imputer.pkl')
    joblib.dump(Xr_train.columns.tolist(), 'rent_feature_names.pkl')
    print("Сохранены rent_model.pkl, rent_imputer.pkl и rent_feature_names.pkl")
```

На этом этапе я подобрала оптимальные гиперпараметры для каждой модели с помощью RandomizedSearchCV — алгоритм перебирает случайные комбинации параметров и оценивает их через кросс-валидацию по метрике MAE. Для линейной модели и дерева решений поиск шёл на полном обучающем наборе, а для случайного леса и градиентного бустинга — на 10%-ной подвыборке, потому что эти модели очень ресурсоёмкие, и полный перебор занял бы часы. В результате Ridge получил максимальную регуляризацию, дерево решений выросло без ограничения глубины, случайный лес взял 200 деревьев глубиной 20, а бустинг — классическую конфигурацию с глубиной 3 и learning rate 0.1. Эти параметры дальше используются для финального обучения и сравнения моделей.

## 3.3 Метрики качества

```python
def calc_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    r2 = r2_score(y_true, y_pred)
    return {'MAE': mae, 'RMSE': rmse, 'MAPE': mape, 'R2': r2}

models = {
    'Linear Regression': best_ridge,
    'Decision Tree': best_dt,
    'Random Forest': best_rf,
    'Gradient Boosting': best_gb
}

res = {}
for name, model in models.items():
    pred = model.predict(X_test_imp)
    res[name] = calc_metrics(y_test, pred)

display(pd.DataFrame(res).T.round(2))
```

На этом этапе я рассчитала четыре метрики качества для каждой модели на тестовой выборке, которую модель не видела при обучении. Результаты показали, что линейная модель Ridge проигрывает всем древовидным — её R² всего 0.42, что подтверждает вывод из EDA о нелинейной зависимости цены от признаков. Лучшей по R² и RMSE оказалась модель Random Forest с показателем 0.54 — она объясняет больше половины дисперсии цен и наиболее устойчива к выбросам. Interestingly, Decision Tree показала наименьший MAE, но её высокий RMSE говорит о том, что она допускает редкие, но очень крупные ошибки. MAPE у всех моделей высокий (29–49%), что объясняется огромным разбросом цен в датасете — от 50 тысяч до миллионов лир. На дешёвых объектах относительная ошибка всегда получается больше. В целом, R² около 0.54 — это разумный результат для задачи прогнозирования недвижимости, где половина ценообразования зависит от факторов, не представленных в данных: точного расположения, вида, состояния ремонта и инфраструктуры района. Для дальнейшего использования в приложении я выбираю Random Forest как наиболее сбалансированную модель
